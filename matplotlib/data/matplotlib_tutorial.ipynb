```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Sample plot


```python
x=[0,1,2,3,4,5]
y=[0,3,6,9,12,15]

plt.title("First graph",fontdict={'fontname':'Times New Roman','fontsize':'20'})
plt.xlabel("X Axis",fontdict={'fontname':'Times New Roman','fontsize':'10'})
plt.ylabel("Y Axis",fontdict={'fontname':'Times New Roman','fontsize':'10'})

plt.plot(x,y)
plt.show()
```

```python
a,b,c=1,-4,1
x=np.arange(-10,11,1)
y=[]

for val in x:
    ans=a*val*val+b*val+c
    y.append(ans)

plt.figure(figsize=(5,5),dpi=100)

plt.title("Quadractic graph!")
plt.plot(x,y,color="#171B18",linestyle='--',marker='.',linewidth=2,markersize=10,markeredgecolor="#E5F62E",label='x2-4x+1')
plt.savefig('quadractic.png',dpi=200)

plt.legend()
plt.show()
```

Bar Charts


```python
chocolates=['Milky Bar','Snickers','Eclairs','KitKat','Munch']
prices=[10,25,5,55,30]
colors=['#949494',"#C43030","#52BD2A","#29B9E0","#280FB8"]

bars=plt.bar(chocolates,prices,color=colors)
patterns=['/','*','.','*','/']

for i,pattern in enumerate(patterns):
    bars[i].set_hatch(pattern)
    bars[i].set_edgecolor('black')
    bars[i].set_linewidth(0.5)

plt.show()
```

Real Life Examples


```python
#Gas prices
gas_prices=pd.read_csv('gas_prices.csv')
print(gas_prices.columns)

plt.title('Gas Prices in USD',fontdict={'fontname':'Times New Roman','fontsize':10})
plt.xlabel('Year')
plt.ylabel('Price',fontdict={'fontname':'Times New Roman','fontsize':10})
plt.xticks(gas_prices['Year'][::3])
for column in gas_prices.columns:
    print(column)
plt.plot(gas_prices['Year'],gas_prices['Australia'],label='Australia',marker='o',linewidth=1,color="#E0D50E")
plt.plot(gas_prices['Year'],gas_prices['Canada'],label='Canada',marker='*',linewidth=1,color="#DE4419")
plt.plot(gas_prices['Year'],gas_prices['France'],label='France',marker='^',linewidth=1,color="#1A22B1")
plt.legend()
plt.show()
```

```python
#Fifa dataset

df_fifa=pd.read_csv('fifa_data.csv')
top10overall=df_fifa.loc[:10,['Name','Overall']].sort_values(['Overall'],ascending=False)
top10potential=df_fifa.loc[:10,['Name','Potential']].sort_values(['Potential'],ascending=False)

plt.figure(figsize=(12,8))
plt.title('Potential v/s Overall')
x=np.arange(len(top10overall['Name']))
width=0.4

bars1=plt.bar(x-width/2,top10overall['Overall'],width,label='Overall',color="#73E119")
bars2=plt.bar(x+width/2,top10potential['Potential'],width,label='Potential',color="#2697E7")

for bar in bars2:
    yval=bar.get_height()
    plt.text(bar.get_x()+bar.get_width()/2,yval,str(yval),ha='center',va='bottom',fontsize=8)

for bar in bars1:
    yval=bar.get_height()
    plt.text(bar.get_x()+bar.get_width()/2,yval,str(yval),ha='center',va='bottom',fontsize=8)
plt.xticks(x,top10overall['Name'],rotation=45,ha='right')
plt.legend(loc='lower right')
plt.show()
```

Histogram

```python
#Nationality
bins=[40,50,60,70,80,90,100]
countries=['England','Germany','Spain','Argentina','France']
for country in countries:
    players=df_fifa.loc[df_fifa['Nationality']==country]
    plt.title(country+' Players Overall')
    counts,bin_edges,patches=plt.hist(players['Overall'],bins=bins,label='Count',color="#4746AD")

    for i in range(len(counts)):
        #Compute the center
        x_center=(bin_edges[i]+bin_edges[i+1])/2;
        y_val=counts[i]

        plt.text(x_center,y_val,str(y_val),ha='center',va='bottom',fontsize=10)
    plt.legend()
    plt.show()
```

Pie Chart

```python
total_players=df_fifa['Name'].count()

plt.title('Bronze v/s Silver v/s Gold')
bronze=df_fifa.loc[df_fifa['Overall']<65].count()[0]
silver=df_fifa.loc[(df_fifa['Overall']>=65) & (df_fifa['Overall']<=74)].count()[0]
gold=df_fifa.loc[df_fifa['Overall']>=75].count()[0]
labels=['Bronze','Silver','Gold']
colors=["#F4A25B","#B5B1B1","#F7D21A"]

plt.pie([bronze,silver,gold],labels=labels,colors=colors,autopct='%.2f',pctdistance=0.5)
plt.show()
```

Box and Whisker Plot

```python
barca=df_fifa.loc[df_fifa['Club']=='FC Barcelona']['Overall']
real_madrid=df_fifa.loc[df_fifa['Club']=='Real Madrid']['Overall']
athletico_madrid=df_fifa.loc[df_fifa['Club']=='Atlético Madrid']['Overall']
man_city=df_fifa.loc[df_fifa['Club']=='Manchester City']['Overall']
man_united=df_fifa.loc[df_fifa['Club']=='Manchester United']['Overall']
chelsea=df_fifa.loc[df_fifa['Club']=='Chelsea']['Overall']
liverpool=df_fifa.loc[df_fifa['Club']=='Liverpool']['Overall']

plt.figure(figsize=(20,10))
plt.title('Club comparisons')
labels=['FC Barcelona','Real Madrid','Atlético Madrid','Manchester City',
        'Manchester United','Chelsea','Liverpool']
x=np.arange(1,len(labels)+1)

plt.boxplot([barca,real_madrid,athletico_madrid,man_city,man_united,
             chelsea,liverpool],label=labels,medianprops={'linewidth':2})

plt.xticks(x,labels)
plt.legend(loc='upper right')
plt.show()
```
