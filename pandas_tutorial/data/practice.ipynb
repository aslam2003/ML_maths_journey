Merging DataFrames

```python
bios=pd.read_csv("bios.csv")
noc_regions=pd.read_csv("noc_regions.csv")
print(bios.columns)
print(noc_regions.columns)
```

```python
filtered_results=df_csv.loc[:,["year","type","discipline","event","athlete_id","medal","team","place","tied"]]
merged_df=pd.merge(filtered_results,bios,left_on='athlete_id',right_on='athlete_id',how='inner')
merged_df=merged_df[merged_df["medal"].notna()].sample(10)
merged_df
```

```python
#heights
merged_df["height_class"]=merged_df["height_cm"].apply(lambda x:"Not Available" if pd.isna(x) else "Short" if x<160.0 else "Moderate" if x<185.0 else "Tall")
merged_df[["name","height_cm","height_class"]].sample(5)
```

```python
pivot=pd.pivot_table(merged_df,values="athlete_id",columns="medal",index="event",aggfunc="count",fill_value=0)
print(pivot)
```
